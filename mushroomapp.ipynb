Upload a photo and click **Classify** to try it.
There are three mushrooms, one edible, one fun one and one deadly.You can upload any of these tree mushrooms **Amanita muscaria**, **Amanita phalloides**, **Russula xerampelina**
> ⚠️ **Never use this to decide whether a mushroom is safe to eat.**

```python
from fastai.vision.widgets import *
```

Now lets indentify the top 3 most dangerous mushrooms. First we search the images

```python
learn_inf=load_learner('export.pkl')
```

```python
learn_inf.dls.vocab
```

```python
def on_click_classify(change):
    img = PILImage.create(btn_upload.data[-1])
    out_pl.clear_output()
    with out_pl: display(img.to_thumb(128,128))
    pred,pred_idx,probs = learn_inf.predict(img)
    lbl_pred.value = f'Prediction: {pred}; Probability: {probs[pred_idx]:.04f}'
    
btn_run = widgets.Button(description='Classify')
btn_run.on_click(on_click_classify)
```

```python
btn_upload = widgets.FileUpload()
out_pl = widgets.Output()
lbl_pred = widgets.Label()
```

```python
VBox([widgets.Label('Select you mushroom'),
      btn_upload,btn_run,out_pl,lbl_pred])
```
